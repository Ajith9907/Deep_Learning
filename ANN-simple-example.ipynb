```python
import numpy as np
import pandas as pd

# 1. Create the dataset
data = {
    "friend_available": [0, 0, 1, 1],
    "weather_good": [0, 1, 0, 1],
    "decision": [0, 1, 1, 1]
}

df = pd.DataFrame(data)

# 2. Extract inputs and target
X = df[["friend_available", "weather_good"]].values
y = df["decision"].values

# 3. Define weights, bias, and activation function
w = np.array([0.6, 0.6])
b = -0.5

def activation(z):
    if z >= 0:
        return 1
    else:
        return 0

# 4. Test for a single situation: Friend is free (1), but weather is bad (0)
test_input = np.array([1, 0])
z = np.dot(test_input, w) + b
prediction = activation(z)

print("--- Single Test ---")
print("Weighted Sum (z):", z)
print("Prediction (0=Stay Home, 1=Go Play):", prediction)

# 5. Predict for all situations and check accuracy
predictions = []
for i in range(len(X)):
    z = np.dot(X[i], w) + b
    predictions.append(activation(z))

df["prediction"] = predictions

correct = np.sum(df["decision"] == df["prediction"])
accuracy = (correct / len(df)) * 100

print("\n--- Full Dataset Evaluation ---")
print(df)
print(f"Accuracy: {accuracy}%")
```
